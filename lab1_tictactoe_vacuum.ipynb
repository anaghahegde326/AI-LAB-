```python
b = [" "] * 9

def show():
    print(f"{b[0]}|{b[1]}|{b[2]}")
    print("-+-+-")
    print(f"{b[3]}|{b[4]}|{b[5]}")
    print("-+-+-")
    print(f"{b[6]}|{b[7]}|{b[8]}")

win = [(0,1,2),(3,4,5),(6,7,8),
       (0,3,6),(1,4,7),(2,5,8),
       (0,4,8),(2,4,6)]

p = "X"

for _ in range(9):
    show()
    n = int(input(f"Player {p}, enter position (1-9): ")) - 1

    if b[n] != " ":
        print("Already occupied!")
        continue

    b[n] = p

    if any(b[a] == b[c] == b[d] == p for a,c,d in win):
        show()
        print(p, "wins!")
        break

    p = "O" if p == "X" else "X"
else:
    show()
    print("Draw!")
```

```python
def reflex(location, status):
    if status == "Dirty":
        return "Suck"
    return "Move Right" if location == "A" else "Move Left"


def goal_agent_n_rooms(rooms, current_location):

    for room_name, status in rooms.items():
        if status == "Dirty":
            if room_name == current_location:
                return "Suck"
            else:
                return f"Move to {room_name}"
    return "Goal Achieved"



print("\n--- Goal-Based Agent for N Rooms ---")

num_rooms = int(input("Enter the number of rooms (e.g., 3): "))
rooms_list = [chr(65 + i) for i in range(num_rooms)]
initial_rooms_status = {}

print("Enter initial status for each room (Clean/Dirty):")
for room_name in rooms_list:
    status = input(f"Room {room_name} Status (Clean/Dirty): ").capitalize()
    initial_rooms_status[room_name] = status

current_location = input(f"Enter initial agent location ({'/'.join(rooms_list)}): ").upper()

while current_location not in rooms_list:
    print("Invalid location. Please choose from available rooms.")
    current_location = input(f"Enter initial agent location ({'/'.join(rooms_list)}): ").upper()



rooms_state = initial_rooms_status.copy()

print("\nStarting Agent Simulation...")

step = 0
while True:
    step += 1
    print(f"\n--- Step {step} ---")
    print(f"Current Rooms State: {rooms_state}")
    print(f"Agent at: {current_location}")

    action = goal_agent_n_rooms(rooms_state, current_location)
    print("Action:", action)

    if action == "Goal Achieved":
        print("All rooms are clean! Goal Achieved.")
        break
    elif action == "Suck":
        rooms_state[current_location] = "Clean"
        print(f"Room {current_location} cleaned.")
    elif action.startswith("Move to "):
        target_room = action.split("Move to ")[1]
        current_location = target_room
        print(f"Moved to Room {current_location}.")
    else:
        print("Unknown action.")

```
