```python

from itertools import product

def implies(p, q):
return (not p) or q


def tt_entails(kb, query):
symbols = ['P', 'Q', 'R']
print("\nTruth Table:")
print("P\tQ\tR\t~Q\tQ->P\tP->~Q\tQ|R\tKB")

result = True

for P, Q, R in product([True, False], repeat=3):
not_Q = not Q

q_implies_p = implies(Q, P)
p_implies_not_q = implies(P, not_Q)
q_or_r = Q or R

# Knowledge Base
KB = q_implies_p and p_implies_not_q

print(
f"{str(P)[0]}\t{str(Q)[0]}\t{str(R)[0]}\t"
f"{str(not_Q)[0]}\t{str(q_implies_p)[0]}\t"
f"{str(p_implies_not_q)[0]}\t{str(q_or_r)[0]}\t"
f"{str(KB)[0]}"
)

# Check entailment
if KB and not q_or_r:
result = False

return result


print("TT-Entails Program")
answer = tt_entails(None, None)

print("\nDoes KB entail Query?")
print("Yes" if answer else "No")
```
