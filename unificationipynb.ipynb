```python
def unify(expr1, expr2):
    # If both expressions are equal
    if expr1 == expr2:
        return {}

    # Split predicates and arguments
    pred1 = expr1[:expr1.index('(')].strip()
    pred2 = expr2[:expr2.index('(')].strip()

    args1 = expr1[expr1.index('(') + 1:expr1.rindex(')')].split(',')
    args2 = expr2[expr2.index('(') + 1:expr2.rindex(')')].split(',')

    args1 = [a.strip() for a in args1]
    args2 = [a.strip() for a in args2]

    # Predicate names must match
    if pred1 != pred2:
        return None

    # Number of arguments must match
    if len(args1) != len(args2):
        return None

    substitution = {}

    # Unify corresponding arguments
    for x, y in zip(args1, args2):
        if x == y:
            continue

        # Variable in first expression
        if x.islower() and x.isalpha():
            if x in substitution:
                if substitution[x] != y:
                    return None
            else:
                substitution[x] = y

        # Variable in second expression
        elif y.islower() and y.isalpha():
            if y in substitution:
                if substitution[y] != x:
                    return None
            else:
                substitution[y] = x

        else:
            return None

    return substitution


# Main program
print("UNIFICATION ALGORITHM")

expr1 = input("Enter first expression: ")
expr2 = input("Enter second expression: ")

result = unify(expr1, expr2)

if result is None:
    print("Unification failed.")
elif not result:
    print("Unification successful.")
    print("Substitution: {}")
else:
    print("Unification successful.")
    print("Substitution:", result)
```
